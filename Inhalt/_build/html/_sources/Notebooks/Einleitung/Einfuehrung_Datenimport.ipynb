# Ausführen von Code in Jupyter Notebooks

Es ist wichtig zu beachten, dass der Code in den Zellen in der Reihenfolge ausgeführt wird, in der sie ausgeführt werden. Wenn also Code in einer bestimmten Zelle geändert und erneut ausgeführt wird, werden die Ergebnisse entsprechend aktualisiert.

# Grundlegende Python-Konzepte

Python bietet eine Vielzahl von eingebauten Funktionen und Konzepten, die das Programmieren erleichtern. Hier sind einige der wichtigsten Grundlagen:

## Variablen und Datentypen

Variablen dienen dazu, Werte zu speichern. Jeder Wert in Python hat einen bestimmten **Datentyp**, der bestimmt, welche Art von Daten gespeichert ist und welche Operationen damit durchgeführt werden können. Die wichtigsten einfachen Datentypen sind:

- **`int`**: Ganze Zahlen (z.B. `10`, `-5`, `0`)
- **`float`**: Gleitkommazahlen (Dezimalzahlen, z.B. `3.14`, `-0.5`, `1.0`)
- **`str`**: Zeichenketten (Text, in Anführungszeichen eingeschlossen, z.B. `'Hallo'`, `"Python"`)
- **`bool`**: Wahrheitswerte (`True` oder `False`)

Mit der Funktion `type()` können wir den Datentyp einer Variablen überprüfen.

````python
# Eine Variable 'a' mit einem Float-Wert
a = 1.1
print(a)
type(a)
````

````python
# Eine Variable 'b' mit einem String-Wert
b = "1.1"
print(b)
type(b)
````

````python
# Eine Variable 'c' mit einem Integer-Wert
c = 1
print(c)
type(c)
````

````python
# Eine Variable 'd' mit einem Boolean-Wert
d = True
print(d)
type(d)
````

## Umwandlung von Datentypen (Type Casting)

Manchmal müssen wir den Datentyp einer Variablen ändern. Zum Beispiel, wenn wir eine Zahl, die als Text gespeichert ist (String), in eine tatsächliche Zahl (Integer oder Float) umwandeln wollen, um damit rechnen zu können. Dies nennt man Type Casting.

- `int()`: Wandelt in eine ganze Zahl um.
- `float()`: Wandelt in eine Gleitkommazahl um.
- `str()`: Wandelt in eine Zeichenkette um.

````python
a = int(1.1)
print(a)
````

````python
type(a)
````

````python
float(a)
````

````python
str(a)
````

## Funktionen

Funktionen sind wiederverwendbare Codeblöcke, die eine bestimmte Aufgabe erfüllen. Sie helfen, Code zu organisieren und Wiederholungen zu vermeiden. Man definiert eine Funktion mit `def` und kann ihr Argumente (Parameter) übergeben.

````python
def square():
    '''
    Diese Funktion berechnet das Quadrat von 2.
    '''
    return 2 * 2
````

````python
square()
````

````python
def square(x):
    '''
    Diese Funktion berechnet das Quadrat einer gegebenen Zahl x.
    
    Parameter:
        x (int oder float): Die Zahl, deren Quadrat berechnet werden soll.
    
    Rückgabewert:
        int oder float: Das Quadrat von x.
    '''
    return x ** 2
````

````python
square16 = square(16)
````

````python
square16
````

````python
str(square16)
````

````python
square(x=3)
````

````python
def quadrat(zahl):
    return zahl ** 2
````

````python
quadrat(zahl=3)
````

1.  **Absoluter Pfad**: Geben Sie den vollständigen Pfad von der Wurzel des Dateisystems an. Dies ist spezifisch für jeden Computer und jedes Betriebssystem. Beispiel (muss angepasst werden!):

Der relative Pfad zum Datenordner ist also `../../Projekt_Daten`. Wir können diesen Pfad mit dem Dateinamen kombinieren, um den vollständigen relativen Pfad zur Datei zu erhalten.
````python
# Pfad zum Ordner, der die Daten enthält (relativ zum aktuellen Verzeichnis)
data_folder_relative = '../../Projekt_Daten/'

# Dateiname
file_name = 'Hidden_Creek_stage_data.csv'

# Kombinieren von Ordnerpfad und Dateiname mit os.path.join()
HC_data_path_relative = os.path.join(data_folder_relative, file_name)

print(f"Relativer Pfad zur Datendatei:\n{HC_data_path_relative}")

# Jetzt können wir die Datei mit dem relativen Pfad einlesen
try:
    df = pd.read_csv(HC_data_path_relative)
    print("\nDatei erfolgreich eingelesen!")
except FileNotFoundError:
    print("\nFehler: Datei nicht unter dem angegebenen Pfad gefunden. Bitte Pfad prüfen!")
````

Es kann so vermieden werden, dass Dinge mehrmals geschrieben oder deklariert werden (mehr Eingaben = höhere Wahrscheinlichkeit eines Tippfehlers). 
Dies sollte grundsätzlich gleich zu Beginn einer Arbeitsdatei geschehen. Es sollten immer leicht verständliche Variablennamen, z. B. `data_folder = <some path to data>`.

>**Hinweis**: Es empfiehlt sich, leicht verständliche Variablennamen zu verwenden. ## Erste Dateninspektion und Qualitätschecks

Nachdem wir die Daten erfolgreich in einen Pandas DataFrame (`df`) geladen haben, ist der nächste wichtige Schritt, einen ersten Blick darauf zu werfen und grundlegende Qualitätschecks durchzuführen. Pandas bietet dafür sehr nützliche Funktionen:

- **`df.head(n)`**: Zeigt die ersten `n` Zeilen des DataFrames an (Standard ist 5). Nützlich, um die Spaltennamen und die ersten Datenpunkte zu sehen.
- **`df.tail(n)`**: Zeigt die letzten `n` Zeilen an.
- **`df.shape`**: Gibt ein Tupel zurück, das die Anzahl der Zeilen und Spalten angibt `(Zeilen, Spalten)`.
- **`df.info()`**: Gibt eine kompakte Zusammenfassung des DataFrames, einschließlich der Anzahl der Zeilen und Spalten, der Spaltennamen, der Anzahl der Nicht-Null-Werte pro Spalte und der Datentypen (`Dtype`) jeder Spalte. **Sehr wichtig, um fehlende Werte und falsche Datentypen schnell zu erkennen!**
- **`df.describe()`**: Berechnet deskriptive Statistiken (Anzahl, Mittelwert, Standardabweichung, Min/Max, Quartile) für numerische Spalten. Hilft, die Verteilung der Daten zu verstehen und Ausreißer zu identifizieren.
- **`df.dtypes`**: Zeigt nur die Datentypen jeder Spalte an.
- **`df.isnull().sum()`**: Zählt die Anzahl der fehlenden Werte (`NaN` - Not a Number) pro Spalte. **Essenziell für die Datenqualität!**

````python
# Zeige die ersten 5 Zeilen
print("--- Erste 5 Zeilen (head) ---")
display(df.head()) # display() wird in Notebooks oft für schönere Tabellenausgabe verwendet

# Zeige die Form (Zeilen, Spalten)
print(f"\n--- Form (Zeilen, Spalten) ---")
print(df.shape)

# Zeige die Info-Zusammenfassung
print("\n--- Info ---")
df.info()
````

Die `df.info()` Ausgabe ist besonders aufschlussreich:
- Wir sehen die Anzahl der Einträge (Zeilen).
- Wir sehen die Spaltennamen und die Anzahl der *nicht fehlenden* (`non-null`) Werte pro Spalte.
- Wir sehen den **Datentyp (`Dtype`)** jeder Spalte.

**Beobachtung:** Die Spalte `Value` hat den Datentyp `object`. In Pandas bedeutet `object` oft, dass die Spalte Zeichenketten (Strings) enthält oder eine Mischung aus verschiedenen Typen ist. Für numerische Berechnungen benötigen wir jedoch einen numerischen Typ wie `float` oder `int`.

Schauen wir uns nun die deskriptiven Statistiken an.

````python
print("\n--- Deskriptive Statistiken (describe) ---")
display(df.describe(include='all')) # include='all' zeigt Statistiken auch für nicht-numerische Spalten
````

Die `describe()`-Ausgabe für die Spalte `Value` bestätigt, dass sie als `object` (wahrscheinlich String) behandelt wird. Wir sehen Statistiken wie `unique` (Anzahl einzigartiger Werte), `top` (häufigster Wert) und `freq` (Häufigkeit des häufigsten Werts), aber keine numerischen Statistiken wie `mean` (Mittelwert).

Dies bedeutet, dass wir keine direkten numerischen Berechnungen mit der Spalte `Value` durchführen können, wie z.B. den Mittelwert berechnen. Ein Versuch würde zu einem Fehler führen.

````python
# Versuch, den Mittelwert der 'Value'-Spalte zu berechnen (wird fehlschlagen)
try:
    mean_value = df['Value'].mean()
except TypeError as e:
    print(f"Fehler beim Berechnen des Mittelwerts (erwartet): {e}")
    print("Grund: Die Spalte 'Value' hat keinen numerischen Datentyp.")
````

Der Fehler (`TypeError: unsupported operand type(s) for +: 'int' and 'str'` oder ähnlich) bestätigt das Problem.

Wenn eine CSV-Datei mit Pandas importiert wird, versucht Pandas automatisch, den Typ jeder Spalte zu erkennen. Wenn eine Spalte jedoch Werte enthält, die nicht eindeutig als Zahlen interpretiert werden können (z.B. Text, gemischte Typen, spezielle Zeichen), weist Pandas ihr oft den allgemeinen `object`-Typ zu.

Dies ist ein **sehr häufiges Problem** bei realen Daten! Ein wichtiger Teil der Datenanalyse besteht darin, solche Typ-Probleme zu identifizieren und zu beheben (Datenbereinigung).

**Niemals** sollten große CSV-Dateien manuell bearbeitet werden, um solche Probleme zu lösen. Wir nutzen Python und Pandas, um dies effizient und reproduzierbar zu tun.

### Zugriff auf Daten im Dataframe

Die Daten wurden bis dato in einer Variablen namens „df“ gespeichert. Die Pandas-Bibliothek ist in vielerlei Hinsicht unglaublich leistungsfähig zum Filtern, Slicen und Indizieren von Daten. Unser Dataframe ist eine Tabelle mit benannten Spalten. Wir können die Namen der Spaltenüberschriften auflisten, indem wir das Attribut „.columns“ des Datenrahmens aufrufen:

````python
df.columns
````

Es kann auf jede beliebige Kombination von Spalten zurückgeriffen und Operationen für einzelne Zeilen oder die gesamte Tabelle ausgeführt werden. Es kann bspw. die Spalte `Value` abgerufen werden, indem wie folgt indiziert wird: `df['Wert']`. Dies gibt einen Array (wie ein Vektor) aller Werte der Spalte zurück.

>**Hinweis**: Wenn auf eine Spalte des Dataframes verwiesen wird, muss diese **genau mit der Zeichenfolge übereinstimmen**. Es ist zu empfehlen, kurze, intuitive Namen zu verwenden und Leerzeichen zu vermeiden. Beim Importieren von Daten von einem Gerät oder aus einer offenen Datenquelle können manchmal Leerzeichen in die Header-Namen eingefügt werden, was zu verwirrenden Fehlern führen kann. d. h. ` Value` ist nicht gleich `Value`.

````python
'Value' == ' Value'
````

Schauen wir uns die Spalte `Value` genauer an und überprüfen den Typ einiger ihrer Elemente.

````python
df['Value']
````

Obwohl die Werte wie Zahlen aussehen, hat uns `df.info()` bereits verraten, dass der *gesamte Spaltentyp* `object` ist. Überprüfen wir den Typ des ersten Elements:

> **Wichtige Erinnerung:** Die Indizierung in Python (und Pandas) beginnt bei **0**. Das erste Element hat den Index `0`, das zweite den Index `1`, usw. `df['Value'][0]` greift also auf den ersten Wert in der Spalte `Value` zu.

````python
# Typ des ersten Elements in der Spalte 'Value'
first_value = df['Value'][0]
print(f"Erster Wert: {first_value}")
print(f"Datentyp des ersten Werts: {type(first_value)}")
````

Der Datentyp ist `str` (String), also Text. Deshalb schlugen die numerischen Operationen fehl. Ein großer Teil der Datenanalyse besteht darin, diese Art von Problemen zu identifizieren und zu beheben! Je schneller und systematischer, desto besser!

### Konvertieren von Datentypen in Pandas

Da wir mit den Wasserstandsdaten numerische Berechnungen durchführen möchten, müssen wir die Spalte `Value` in einen numerischen Datentyp umwandeln. Da Wasserstände Dezimalwerte haben können, ist `float` (Gleitkommazahl) der geeignete Typ.

Pandas bietet dafür die Methode `.astype()`:
`df['NeueSpalte'] = df['AlteSpalte'].astype(float)`

Es ist gute Praxis, der neuen, konvertierten Spalte einen aussagekräftigeren Namen zu geben, der idealerweise auch die Einheit enthält (wenn bekannt). Wir nennen sie `stage_m` (für Wasserstand in Metern).

**Alternative:** Pandas bietet auch die Funktion `pd.to_numeric()`. Ein Vorteil davon ist die Option `errors='coerce'`, die automatisch alle Werte, die nicht in Zahlen umgewandelt werden können, durch `NaN` (Not a Number - fehlender Wert) ersetzt. Das kann sehr nützlich sein, um problematische Einträge zu finden.
`df['stage_m'] = pd.to_numeric(df['Value'], errors='coerce')`

Für dieses Beispiel verwenden wir `.astype(float)`. Wenn dies fehlschlägt, wissen wir, dass es nicht-numerische Einträge gibt, die wir untersuchen müssten. Wenn es funktioniert, waren alle Strings als Zahlen interpretierbar.

````python
try:
    # Versuche, die 'Value'-Spalte in float zu konvertieren und in neuer Spalte zu speichern
    df['stage_m'] = df['Value'].astype(float)
    print("Konvertierung zu float erfolgreich.")
    
    # Überprüfe den Datentyp der neuen Spalte
    print(f"Datentyp der neuen Spalte 'stage_m': {df['stage_m'].dtype}")
    
    # Zeige die ersten Werte der neuen Spalte
    print("\nErste Werte der neuen Spalte 'stage_m':")
    display(df[['Value', 'stage_m']].head())
    
except ValueError as e:
    print(f"Fehler bei der Konvertierung (ValueError): {e}")
    print("Dies bedeutet, dass mindestens ein Wert in der Spalte 'Value' nicht direkt in eine Zahl umgewandelt werden kann.")
    print("In diesem Fall wäre pd.to_numeric(df['Value'], errors='coerce') eine gute Alternative.")
````

Der Datentyp wurde erfolgreich zum Typ `float` geändert.

Mit Gleitkomma wird die Art und Weise beschrieben, wie Computer Brüche darstellen. In der [Python-Dokumentation](https://docs.python.org/3/tutorial/floatingpoint.html) wird die Zahl „0,125“ als $\frac{1}{10}+\frac{2}{100}+\frac{5}{1000}$ dargestellt. Die meisten Zahlen können auf diese Weise nicht perfekt dargestellt werden, weil wir kein unendliches Gedächtnis haben (denken Sie zum Beispiel an Pi oder den Bruch „1/3“).

Jetzt können wir den mittleren Wasserstand berechnen:

````python
# Jetzt können wir den mittleren Wasserstand berechnen:
mean_stage = df['stage_m'].mean()
print(f"Mittlerer Wasserstand: {mean_stage:.2f} m") # :.2f formatiert auf 2 Nachkommastellen
````

### Umgang mit fehlenden Werten (`NaN`) und Visualisierung

Ein weiterer wichtiger Aspekt der Datenqualität sind **fehlende Werte**. Diese werden in Pandas oft als `NaN` (Not a Number) dargestellt.

Wir können die Anzahl der fehlenden Werte pro Spalte mit `df.isnull().sum()` überprüfen.

````python
print("Anzahl fehlender Werte pro Spalte:")
print(df.isnull().sum())
````

**Beobachtung:** Die Spalte `stage_m` enthält fehlende Werte (`NaN`). Das ist interessant, da die `.astype(float)` Konvertierung funktioniert hat. Das bedeutet, dass die ursprüngliche CSV-Datei wahrscheinlich leere Einträge oder spezielle Zeichenketten enthielt, die Pandas beim Einlesen standardmäßig als `NaN` interpretiert hat, *bevor* wir die Typkonvertierung durchgeführt haben.

Fehlende Werte können Analysen beeinflussen und müssen oft behandelt werden (z.B. durch Entfernen der Zeilen, Interpolation, Auffüllen mit einem Mittelwert etc.).

Eine **Visualisierung** der Daten als Zeitreihe ist oft der beste Weg, um Muster, Trends, Ausreißer und eben auch Datenlücken schnell zu erkennen.

Für Zeitreihenplots ist es sehr hilfreich, die Zeitspalte als **Index** des DataFrames zu verwenden. Pandas erkennt Datums- und Zeitangaben oft automatisch, aber manchmal müssen wir sie explizit mit `pd.to_datetime()` umwandeln. Ein `DatetimeIndex` ermöglicht spezielle Zeitreihenoperationen und schönere Achsenbeschriftungen in Plots.

````python
# Wandle die 'Date'-Spalte in Datetime-Objekte um und setze sie als Index
# 'inplace=True' modifiziert den DataFrame direkt, ohne eine Kopie zurückzugeben.
try:
    df.index = pd.to_datetime(df['Date'])
    # Optional: Entferne die ursprüngliche 'Date'-Spalte, da sie jetzt der Index ist
    # df.drop('Date', axis=1, inplace=True)
    print("'Date'-Spalte erfolgreich als DatetimeIndex gesetzt.")
    display(df.head(3))
except Exception as e:
    print(f"Fehler beim Setzen des DatetimeIndex: {e}")
````

Jetzt können wir die Wasserstandsdaten (`stage_m`) gegen den Zeitindex plotten.

````python
# Erstelle eine Abbildung und eine Achse mit matplotlib
fig, ax = plt.subplots(figsize=(12, 5)) # Größe der Abbildung anpassen

# Plotte die 'stage_m'-Spalte gegen den Index (Zeit)
ax.plot(df.index, df['stage_m'], label='Wasserstand [m]', marker='.', linestyle='-', markersize=2, alpha=0.7)

# Achsenbeschriftungen und Titel
ax.set_xlabel('Datum', size=12)
ax.set_ylabel('Wasserstand [m]', size=12)
ax.set_title('Zeitreihe Wasserstand Hidden Creek', size=14)

# Legende anzeigen
ax.legend()

# Gitternetz hinzufügen (optional)
ax.grid(True, linestyle='--', alpha=0.6)

# Layout optimieren und Plot anzeigen
fig.tight_layout()
plt.show()
````

Im Plot sind deutliche **Lücken** zu erkennen, die den zuvor festgestellten `NaN`-Werten entsprechen. Wir können die Zeilen mit fehlenden Werten auch explizit filtern:

````python
# Filtere alle Zeilen, bei denen 'stage_m' NaN ist
gaps = df[df['stage_m'].isnull()]

print(f"Anzahl der Zeilen mit fehlenden Wasserstandsdaten: {len(gaps)}")
if not gaps.empty:
    print("\nErste paar Zeilen mit Lücken:")
    display(gaps.head())
````

Wir haben also eine signifikante Anzahl von Tagen mit fehlenden Daten.

Je nach Anwendungsfall gibt es verschiedene Strategien, mit fehlenden Daten umzugehen:
- **Ignorieren/Entfernen**: Wenn nur wenige Daten fehlen oder die Analyse dies zulässt.
- **Interpolation**: Fehlende Werte durch Schätzungen basierend auf benachbarten Werten ersetzen (z.B. lineare Interpolation).
- **Auffüllen**: Fehlende Werte durch einen festen Wert ersetzen (z.B. Mittelwert, Median).

Hier zeigen wir beispielhaft die **lineare Interpolation** mit der `.interpolate()`-Methode von Pandas.

````python
# Erstelle eine Kopie der Spalte, um das Original nicht zu verändern
df['stage_m_interpolated'] = df['stage_m'].interpolate(method='linear')

# Überprüfe, ob noch NaN-Werte vorhanden sind (sollten keine mehr sein, außer am Anfang/Ende)
print("Fehlende Werte nach Interpolation:")
print(df['stage_m_interpolated'].isnull().sum())
````

````python
# Plotte Originaldaten und interpolierte Daten zum Vergleich
fig, ax = plt.subplots(figsize=(12, 5))

ax.plot(df.index, df['stage_m'], label='Original (mit Lücken)', marker='.', linestyle='None', markersize=3, color='blue', alpha=0.6)
ax.plot(df.index, df['stage_m_interpolated'], label='Linear interpoliert', linestyle='-', color='red', alpha=0.8)

ax.set_xlabel('Datum', size=12)
ax.set_ylabel('Wasserstand [m]', size=12)
ax.set_title('Vergleich: Original vs. Interpolierte Wasserstandsdaten', size=14)
ax.legend()
ax.grid(True, linestyle='--', alpha=0.6)

fig.tight_layout()
plt.show()
````
