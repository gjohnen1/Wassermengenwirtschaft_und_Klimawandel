## 1. Variablen und Datentypen

Eine **Variable** ist ein Name für einen Wert. Die vier wichtigsten einfachen Datentypen:

| Typ | Zweck | Beispiel |
|---|---|---|
| `int` | ganze Zahl | `anzahl_pegel = 7` |
| `float` | Dezimalzahl | `mittlerer_abfluss_m3s = 12.4` |
| `str` | Text (in Anführungszeichen) | `pegelname = "Hidden Creek"` |
| `bool` | Wahrheitswert | `ist_hochwasser = True` |

Mit `type(...)` fragt man den Typ einer Variable ab. Einheiten gehören bei uns konventionell in den Variablennamen (`..._m3s`, `..._m`) — das ist kein Python-Zwang, aber guter Stil und erspart später Einheitenfehler.


````python
# Vier Variablen mit den vier Grundtypen
anzahl_pegel = 7                          # int
mittlerer_abfluss_m3s = 12.4              # float
pegelname = "Hidden Creek"                # str
ist_hochwasser = False                    # bool

for name, wert in [("anzahl_pegel", anzahl_pegel),
                   ("mittlerer_abfluss_m3s", mittlerer_abfluss_m3s),
                   ("pegelname", pegelname),
                   ("ist_hochwasser", ist_hochwasser)]:
    print(f"{name:25s} = {wert!r:15s}  → {type(wert).__name__}")
````

## 2. Type-Casting — Typen umwandeln

Manchmal kommt ein Wert in der „falschen" Form an. Klassisches Beispiel: eine Zahl ist als Text (`str`) gespeichert und lässt sich nicht rechnen. Dann castet man:

- `int(x)` — nach Ganzzahl (schneidet Nachkommastellen **ab**, rundet nicht!)
- `float(x)` — nach Dezimalzahl
- `str(x)` — nach Text


````python
# Beispiel: Wasserstand als String aus einer CSV — in float umwandeln
wasserstand_str = "1.47"
wasserstand_float = float(wasserstand_str)
print(f"{wasserstand_str!r:10s} (Typ {type(wasserstand_str).__name__}) "
      f"→ {wasserstand_float} (Typ {type(wasserstand_float).__name__})")

# int() schneidet ab, rundet nicht:
print(f"int(1.7) = {int(1.7)}     <-- NICHT 2!")
````

## 3. Funktionen

Eine **Funktion** kapselt wiederverwendbaren Code. Syntax: `def name(parameter): ...`. Mit `return` gibt die Funktion ein Ergebnis zurück.

Faustregel: sobald Ihr denselben Code zum dritten Mal schreibt, ist eine Funktion überfällig.


````python
def abfluss_aus_wasserstand(stage_m, a=2.5, b=1.6):
    '''
    Potenzgesetz-Pegelschlüssel: Q = a * (stage)^b

    Parameter:
        stage_m (float): Wasserstand in Metern.
        a, b (float):    Kalibrierungsparameter (Default: Beispielwerte).

    Rückgabe:
        float: Abfluss in m³/s.
    '''
    return a * stage_m ** b

# Aufruf
q = abfluss_aus_wasserstand(1.4)
print(f"Bei stage = 1.4 m → Q ≈ {q:.2f} m³/s")

# Mit benannten Argumenten — genauso gültig, aber lesbarer:
q2 = abfluss_aus_wasserstand(stage_m=1.4, a=3.0, b=1.5)
print(f"Mit anderen Parametern → Q ≈ {q2:.2f} m³/s")
````

## 4. Kontrollfluss — `if` / `else`

Mit `if` führt Ihr Code nur dann aus, wenn eine Bedingung wahr ist. Achtet auf den Doppelpunkt und die **Einrückung** (vier Leerzeichen). Python erzwingt die Einrückung syntaktisch.


````python
abfluss_m3s = 185.0
warnschwelle_m3s = 150.0

if abfluss_m3s > warnschwelle_m3s:
    print(f"WARNUNG: Abfluss {abfluss_m3s:.0f} m³/s "
          f"über Schwelle ({warnschwelle_m3s:.0f} m³/s).")
else:
    print(f"OK: Abfluss {abfluss_m3s:.0f} m³/s unter Schwelle.")
````

## 5. Listen, `for`-Schleifen und Dictionaries

Für die späteren Übungen braucht Ihr drei Strukturen:

- **Liste** (`[...]`) — geordnete Sammlung, indiziert ab `0`.
- **`for`-Schleife** — „mache etwas für jedes Element in einer Liste".
- **Dictionary** (`{...}`) — Zuordnung `Schlüssel → Wert`, z.B. Pegelname → Einzugsgebietsgröße.


````python
# Eine Liste aus Tages-Wasserständen [m]
wasserstaende_m = [1.12, 1.18, 1.23, 1.31, 1.47, 1.55, 1.48]

print(f"Anzahl Messungen: {len(wasserstaende_m)}")
print(f"Erster Wert:      {wasserstaende_m[0]}")
print(f"Letzter Wert:     {wasserstaende_m[-1]}")

# for-Schleife: Abfluss aus jedem Wasserstand schätzen
print("\nBerechnete Abflüsse:")
for i, stage in enumerate(wasserstaende_m):
    q = abfluss_aus_wasserstand(stage)      # Funktion von oben wiederverwendet
    print(f"  Tag {i+1}: stage = {stage:.2f} m  →  Q ≈ {q:.2f} m³/s")
````

````python
# Dictionary: Pegel → Einzugsgebietsgröße [km²]
pegel = {
    "Hidden Creek":   18.4,
    "Chilliwack":    1230.0,
    "Fraser River": 217000.0,
}

# Zugriff per Schlüssel
print(f"Einzugsgebiet Hidden Creek: {pegel['Hidden Creek']} km²")

# Über alle Einträge iterieren
print("\nAlle Pegel:")
for name, flaeche_km2 in pegel.items():
    print(f"  {name:15s} → {flaeche_km2:>10.1f} km²")
````

### Pfade sauber bauen — und das DRY-Prinzip

Gemäß dem Prinzip [**Don't Repeat Yourself**](https://en.wikipedia.org/wiki/Don%27t_repeat_yourself) speichert man Pfade in Variablen — einmal oben im Notebook, danach nur noch über den Variablennamen referenziert. Mehrmaliges Tippen = höhere Fehleranfälligkeit.

Variablennamen sollten **aussagekräftig** sein und Einheiten *nicht* verschleiern. Wenn Ihr einen mittleren Jahresabfluss speichert, ist `qa` heute verständlich, aber in einem Jahr niemandem mehr klar — `avg_annual_flow_m3s` enthält den Zweck *und* die Einheit (m³/s ↔ „cubic metres per second"). **Einheiten sind wichtig.**

### Der Fehler, der zeigt, warum Typen wichtig sind


````python
# Versuch, den Mittelwert der 'Value'-Spalte zu berechnen — schlägt fehl
try:
    mean_value = df['Value'].mean()
    print(f"Mittelwert: {mean_value}")
except TypeError as e:
    print(f"Fehler beim Berechnen des Mittelwerts (erwartet): {e}")
    print("→ Grund: Die Spalte 'Value' hat keinen numerischen Datentyp.")
````

Der Fehler (`TypeError` oder ähnlich) bestätigt: Die `Value`-Spalte ist *Text*, obwohl sie wie eine Zahl aussieht. Das ist ein klassisches Symptom von *kleinen* Darstellungsproblemen in der Quelldatei (Dezimaltrenner, Leerzeichen, Fehlermarkierungen wie `"—"` oder `"NA"`).

### Datentyp konvertieren

Da wir mit den Wasserständen rechnen wollen, brauchen wir einen numerischen Typ — `float`, weil Wasserstände Dezimalstellen haben.

Zwei Wege:

- `df['neu'] = df['alt'].astype(float)` — schnell, aber wirft `ValueError`, wenn *ein einziger* Wert nicht konvertierbar ist.
- `df['neu'] = pd.to_numeric(df['alt'], errors='coerce')` — ersetzt Problemwerte durch `NaN` (*Not a Number*). Robuster und oft die bessere Wahl.

Gute Praxis: der neuen Spalte einen aussagekräftigen Namen *mit Einheit* geben — `stage_m`.


````python
try:
    df['stage_m'] = df['Value'].astype(float)
    print("Konvertierung zu float erfolgreich.")
    print(f"Datentyp der neuen Spalte 'stage_m': {df['stage_m'].dtype}")
    print("\nErste Werte:")
    display(df[['Value', 'stage_m']].head())
except ValueError as e:
    print(f"Fehler (ValueError): {e}")
    print("→ pd.to_numeric(df['Value'], errors='coerce') wäre eine robuste Alternative.")
````

````python
# Jetzt können wir den mittleren Wasserstand berechnen:
mean_stage = df['stage_m'].mean()
print(f"Mittlerer Wasserstand: {mean_stage:.2f} m")
````

### Fehlende Werte und Visualisierung

Die Konvertierung ging glatt — aber das heißt nicht, dass die Daten lückenlos sind. `isnull().sum()` zählt die Anzahl fehlender Werte pro Spalte:


````python
print("Anzahl fehlender Werte pro Spalte:")
print(df.isnull().sum())
````

**Beobachtung:** Die Spalte `stage_m` enthält `NaN`-Werte. `isnull().sum()` ist der schnellste Weg, Datenlücken zu erkennen.

Für Zeitreihen-Plots setzen wir die `Date`-Spalte als Index:


````python
# Date-Spalte in Datetime-Objekte umwandeln und als Index setzen
try:
    df.index = pd.to_datetime(df['Date'])
    print("'Date'-Spalte erfolgreich als DatetimeIndex gesetzt.")
    display(df.head(3))
except Exception as e:
    print(f"Fehler beim Setzen des DatetimeIndex: {e}")
````

````python
# Zeitreihe plotten
fig, ax = plt.subplots(figsize=(12, 5))

ax.plot(df.index, df['stage_m'], label='Wasserstand [m]',
        marker='.', linestyle='-', markersize=2, alpha=0.7)

ax.set_xlabel('Datum', size=12)
ax.set_ylabel('Wasserstand [m]', size=12)
ax.set_title('Zeitreihe Wasserstand Hidden Creek', size=14)
ax.legend()
ax.grid(True, linestyle='--', alpha=0.6)

fig.tight_layout()
plt.show()
````

Im Plot sind **Lücken** erkennbar, die zu den `NaN`-Werten passen. Wir zählen sie gezielt:


````python
gaps = df[df['stage_m'].isnull()]

print(f"Anzahl Zeilen mit fehlenden Wasserstandsdaten: {len(gaps)}")
if not gaps.empty:
    print("\nErste Zeilen mit Lücken:")
    display(gaps.head())
````

### Lücken schließen: lineare Interpolation

Für eine durchgehende Zeitreihe bietet `pandas` die Methode `.interpolate()`. `method='linear'` zieht zwischen den bekannten Punkten eine Gerade. **Wichtig:** Interpolation ist eine *Annahme* über die fehlenden Werte — keine Messung. Für Hochwasseranalysen solltet Ihr sehr vorsichtig damit sein; für Plots ist sie meist okay.


````python
# Original nicht überschreiben — neue Spalte mit interpolierten Werten
df['stage_m_interpolated'] = df['stage_m'].interpolate(method='linear')

print("Fehlende Werte nach Interpolation:")
print(df['stage_m_interpolated'].isnull().sum())
````

````python
# Original (Punkte) und interpolierte Version (Linie) im Vergleich
fig, ax = plt.subplots(figsize=(12, 5))

ax.plot(df.index, df['stage_m'],
        label='Original (mit Lücken)',
        marker='.', linestyle='None', markersize=3, color='blue', alpha=0.6)
ax.plot(df.index, df['stage_m_interpolated'],
        label='Linear interpoliert',
        linestyle='-', color='red', alpha=0.8)

ax.set_xlabel('Datum', size=12)
ax.set_ylabel('Wasserstand [m]', size=12)
ax.set_title('Vergleich: Original vs. interpolierte Wasserstandsdaten', size=14)
ax.legend()
ax.grid(True, linestyle='--', alpha=0.6)

fig.tight_layout()
plt.show()
````

### Drei Prompts, drei Qualitäten

**Warum Prompt 3 besser ist:**

## 4. Der Verifikationsworkflow — warum das alles zählt

Jeder Vorschlag, egal ob Inline oder Chat, durchläuft bei Euch vier Schritte:

1. **Lesen** — versteht Ihr, was der Code tut? Bei jedem Aufruf: welche Funktion, welche Bibliothek, welcher Rückgabewert?
2. **Ausführen** — im Notebook in einer eigenen Zelle. **Nie** direkt in produktivem Code-Abschnitt einfügen.
3. **Prüfen** — *hier* wird es hydrologisch:
   - `df.shape` vor- und nachher: passt die Zeilenzahl?
   - `df.describe()`: sind Minimum/Maximum/Mittelwert plausibel?
   - **Einheiten**: rechnet der Code in m oder cm, in m³/s oder cfs?
   - Plot: sehen die Werte so aus, wie sie sollten?
4. **Übernehmen oder verwerfen** — erst danach in die Hauptzellen Eures Notebooks.

Beispiel: ein LLM schlägt vor, `df.dropna()` statt `df.interpolate()` zu verwenden. Beides ist Code, der läuft. Der Unterschied:


````python
# Check: wie viele Zeilen verliere ich durch dropna() im Vergleich zu interpolate()?
n_original      = len(df)
n_after_dropna  = len(df.dropna(subset=['stage_m']))
n_missing       = df['stage_m'].isnull().sum()

print(f"Zeilen gesamt:                    {n_original}")
print(f"Fehlend in 'stage_m':             {n_missing}")
print(f"Zeilen nach dropna(subset=...):   {n_after_dropna}")
print(f"Verhältnis: {n_after_dropna/n_original:.1%} der Originaldaten bleiben übrig.")
````

Wenn Euch diese Ausgabe sagt, dass `dropna` z.B. 20 % der Zeitreihe löscht, dann wisst Ihr: das ist kein gleichwertiger Ersatz für Interpolation — das ist eine *andere Entscheidung mit anderen Konsequenzen*. Genau das ist die Prüfung, die KI nicht übernehmen kann.

## 6. Halluzinationen — wenn KI eine Funktion erfindet

LLMs erfinden gelegentlich Python-Funktionen, die es gar nicht gibt. Ein reales Beispiel, das Studierenden immer wieder passiert:

> „Nutze `df.interpolate_nans(method='cubic')` für die Lückenfüllung."

Diese Funktion **existiert nicht** in `pandas`. Richtig wäre `df.interpolate(method='cubic')`. Woran Ihr eine Halluzination erkennt:

- Die Methode ist in der [offiziellen pandas-Dokumentation](https://pandas.pydata.org/docs/) nicht zu finden.
- Im Notebook wird beim Ausführen ein `AttributeError: 'DataFrame' object has no attribute '...'` geworfen.
- Der Parameter (`method=...`) wird in einer Variante vorgeschlagen, die die Originalfunktion gar nicht unterstützt.

**Gegenmittel:** Bei jedem neuen Funktionsnamen, den Euch ein LLM vorschlägt, einmal kurz die Doku oder StackOverflow öffnen. Das dauert 20 Sekunden und spart Euch später Stunden Fehlersuche.


## 7. Datenschutz und Vertraulichkeit

- **Keine personenbezogenen Daten** in Prompts (weder Eure noch die von Anderen).
- **Projektdaten unter NDA / vertrauliche Pegeldaten** dürfen nicht in ein Chat-LLM eingegeben werden. Copilot (mit Business-/Enterprise-Lizenz) ist hier strenger; die kostenlose Studenten-Version und Chat-LLMs sind das **nicht** — behandelt sie wie ein öffentliches Forum.
- Bei Unsicherheit: **abstrahiert** (ersetzt Pegelnamen, Orte, konkrete Projektbezüge durch Platzhalter). Ein LLM kann Euch bei einem Datenqualitätsproblem genauso gut helfen, wenn die Spalte im Prompt `stage_m` statt `Pegel_KlientXY_vertraulich` heißt.
- Für die Hausarbeit in diesem Kurs dürft Ihr KI frei nutzen — dokumentiert aber in einem kurzen Satz, wo und wie.
